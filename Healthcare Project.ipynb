```python
# Strip extra spaces in column names
```

```python
df.columns = df.columns.str.strip()
```

```python
# Change the column name and relplace ' ' with '_'
```

```python
df.rename(columns={'Name': 'Patient Name'}, inplace=True)
```

```python
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.dtypes
```

```python
# Change the formate of Admission Date and Discharge Date in YYYY-MM-DD
```

```python
df['Date_of_Admission'] = pd.to_datetime(df['Date_of_Admission']).dt.strftime('%Y-%m-%d')
```

```python
df['Discharge_Date'] = pd.to_datetime(df['Discharge_Date']).dt.strftime('%Y-%m-%d')
```

```python
# Change the data type from object to date for both the column
```

```python
date_cols = ['Date_of_Admission', 'Discharge_Date']

for col in date_cols:
    df[col] = pd.to_datetime(df[col])
```

```python
df.info()
```

```python
df.head()
```

```python
# change Alzheimer’s into Alzheimer in medical_condition column
```

```python
df['Medical_Condition'] = df['Medical_Condition'].replace(
    "Alzheimer’s",
    "Alzheimer"
)
```

```python
df.head(20)
```

```python
# create new column for diseases risk category
```

```python
df['risk_category'] = df['Medical_Condition'].apply(
    lambda x:
    'Critical' if x in ['Cancer', 'Heart Disease', 'Alzheimer']
    else 'High Risk' if x in ['Diabetes', 'Obesity', 'Asthma']
    else 'Low Risk'
)
```

```python
df['age_group'] = df['Age'].apply(
    lambda x:
    'Child' if x <= 18
    else 'Young Adult' if x <= 35
    else 'Adult' if x <= 50
    else 'Senior Adult' if x <= 65
    else 'Elderly'
)
```
